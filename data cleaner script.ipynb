```python
# 19MAcademy — Project 1: Data Cleaner Script

def clean_data(raw_data):

    cleaned = []

    for item in raw_data:

        # Skip missing values
        if item is None:
            continue

        try:
            # Convert values like "85" into numbers
            value = float(item)

            # Skip negative numbers
            if value >= 0:
                cleaned.append(value)

        except (ValueError, TypeError):
            # Skip invalid values like strings
            continue

    # Remove duplicates while keeping order
    unique_cleaned = list(dict.fromkeys(cleaned))

    return unique_cleaned


def analyse(marks):

    if not marks:
        print("No valid data to analyse.")
        return

    print(f"Count   : {len(marks)}")
    print(f"Average : {sum(marks) / len(marks):.2f}")
    print(f"Highest : {max(marks)}")
    print(f"Lowest  : {min(marks)}")


# --- Run Program ---

raw_marks = [80, 90, None, 75, 90, "85", None, 60, -5, "invalid", 75]

print("Raw Data:", raw_marks)

final_marks = clean_data(raw_marks)

print("Clean Data:", final_marks)

print()

analyse(final_marks)
```
